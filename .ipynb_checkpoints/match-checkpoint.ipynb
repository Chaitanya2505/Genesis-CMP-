```python
import numpy as np
import pandas as pd
```

```python
matches = pd.read_csv('/Users/chaitanyaaggarwal/Documents/Genesis-CMP-/github/matches.csv')
```

```python
matches.head()
```

```python
print(matches.columns)
```

```python
# Rename columns to match your desired format
matches = matches.rename(columns={
    'id': 'match_id',
    'date': 'match_date',
    'team1': 'team_1',
    'team2': 'team_2'
})
```

```python
# Logic: If toss winner chose to bat, they are batting_first. 
# Otherwise, the other team (team_2 if team_1 won toss) is batting_first.
matches['batting_first'] = np.where(
    ((matches['toss_winner'] == matches['team_1']) & (matches['toss_decision'] == 'bat')) |
    ((matches['toss_winner'] == matches['team_2']) & (matches['toss_decision'] == 'field')),
    matches['team_1'], 
    matches['team_2']
)
```
