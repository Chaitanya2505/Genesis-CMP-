```python
import numpy as np
import pandas as pd
```

```python
players = pd.read_csv('Players.csv')
```

```python
players.head()
```

```python
players.shape
```

```python
players = players.drop(columns = ['TEAM','AGE','SOLD_PRICE'])
```

```python
players.head()
```

```python
players
```

```python
# inserts 'Player_ID' at index 0 (the very first column)
players.insert(0, 'Player_ID', range(1, len(players) + 1))
```

```python
players.head()
```

```python
print(players.columns)
```

```python
keep_cols = [
    # Identity
    'Player_ID',
    'Player',
    'Paying_Role',

    # Batting – career
    'Mat', 'Inns', 'Runs', 'Avg', 'SR',

    # Batting – recent (T)
    'TMat', 'TInns', 'TRuns', 'TAvg', 'TSR',

    # Bowling – career
    'B_Inns', 'B_Balls', 'B_Wkts', 'B_Econ',

    # Bowling – recent (T)
    'B_TInns', 'B_TBalls', 'B_TWkts', 'B_TEcon'
]
```

```python
players_clean = players[keep_cols].copy()
```

```python
players_clean.head()
```

```python
players_clean.shape
```

```python
# Fill NaNs with 0 for numeric columns
num_cols = players_clean.select_dtypes(include=['float64', 'int64']).columns
players_clean[num_cols] = players_clean[num_cols].fillna(0)
```

```python
players_clean.head()
```

```python
players_clean['Paying_Role'] = players_clean['Paying_Role'].str.lower()

players_clean['Paying_Role'] = players_clean['Paying_Role'].replace({
    'batting': 'batsman',
    'bowling': 'bowler',
    'all rounder': 'allrounder',
    'allrounder': 'allrounder'
})
```

```python
players_clean.info()
```

```python
players_clean['Avg'] = pd.to_numeric(players_clean['Avg'], errors='coerce')
```

```python
players_clean['Avg'] = players_clean['Avg'].fillna(0)
```

```python
players_clean.info()
```

```python
players_clean = players_clean.rename(columns={'Paying_Role': 'Playing_Role'})
```

```python
players_clean['Playing_Role'] = players_clean['Playing_Role'].astype('category')
```

```python
players_clean.to_csv("player_stats_clean.csv", index=False)
```
