```python
matches_df = pd.read_csv("/Users/chaitanyaaggarwal/Documents/Genesis-CMP-/github/match_level_dataset2.csv")
teams_df = pd.read_csv("/Users/chaitanyaaggarwal/Documents/Genesis-CMP-/github/team_features_normalized.csv")
```

```python
df = matches_df.merge(
    teams_df,
    left_on="team_1",
    right_on="team",
    how="left"
)

df = df.rename(
    columns={col: f"t1_{col}" for col in teams_df.columns if col != "team"}
)

df.drop(columns=["team"], inplace=True)
```

```python
df = df.merge(
    teams_df,
    left_on="team_2",
    right_on="team",
    how="left"
)

df = df.rename(
    columns={col: f"t2_{col}" for col in teams_df.columns if col != "team"}
)

df.drop(columns=["team"], inplace=True)
```

```python
assert df.isnull().sum().sum() == 0, "Nulls found after merging!"
print("✅ Merge successful, dataset ready")
```

```python
target_winner = "winner"
target_score_1 = "first_innings_score"
target_score_2 = "second_innings_score"

categorical_cols = [
    "team_1", "team_2", "toss_winner", "toss_decision", "batting_first"
]

numeric_cols = [col for col in df.columns 
                if col.startswith("t1_") or col.startswith("t2_")]
```

```python
from sklearn.preprocessing import LabelEncoder

encoders = {}
for col in categorical_cols + [target_winner]:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])
    encoders[col] = le
```

```python
X = df[categorical_cols + numeric_cols]
y_winner = df[target_winner]
y_score_1 = df[target_score_1]
y_score_2 = df[target_score_2]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y_winner, test_size=0.2, random_state=42
)
```

```python
from sklearn.linear_model import LogisticRegression

winner_model = LogisticRegression(max_iter=1000)
winner_model.fit(X_train, y_train)
```

```python
from sklearn.linear_model import LinearRegression

score_model_1 = LinearRegression()
score_model_2 = LinearRegression()

score_model_1.fit(X, y_score_1)
score_model_2.fit(X, y_score_2)
```

```python
import joblib

joblib.dump(winner_model, "winner_model.pkl")
joblib.dump(score_model_1, "score_model_1.pkl")
joblib.dump(score_model_2, "score_model_2.pkl")
joblib.dump(encoders, "encoders.pkl")
```
