```python
team_name_mapping = {
    "Kings XI Punjab": "Punjab Kings",
    "Delhi Daredevils": "Delhi Capitals",
    "Deccan Chargers": "Sunrisers Hyderabad",
    "Royal Challengers Bangalore": "Royal Challengers Bengaluru"
}
```

```python
for col in ["team_1", "team_2", "toss_winner", "batting_first", "winner"]:
    matches[col] = matches[col].replace(team_name_mapping)
```

```python
valid_teams = [
    "Chennai Super Kings",
    "Delhi Capitals",
    "Gujarat Titans",
    "Kolkata Knight Riders",
    "Lucknow Super Giants",
    "Mumbai Indians",
    "Punjab Kings",
    "Rajasthan Royals",
    "Royal Challengers Bengaluru",
    "Sunrisers Hyderabad"
]

matches = matches[
    matches["team_1"].isin(valid_teams) &
    matches["team_2"].isin(valid_teams)
].reset_index(drop=True)
```
