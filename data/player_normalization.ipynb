```python
import pandas as pd

df = pd.read_csv("player_stats_normalized.csv")

# ----- Batting Rating -----
df["batting_rating"] = (
    0.6 * df["TAvg"].fillna(0) +
    0.4 * (df["TSR"].fillna(0) / 150) * 100
)

# ----- Bowling Rating -----
df["bowling_rating"] = (
    (df["B_TWkts"].fillna(0) / df["B_TInns"].replace(0, 1)) * 15 +
    (10 - df["B_TEcon"].fillna(10)) * 6
).clip(0, 100)

# ----- Form Rating -----
df["form_rating"] = (
    df["batting_rating"] * 0.5 +
    df["bowling_rating"] * 0.5
)

out = df[[
    "Player",
    "Playing_Role",
    "batting_rating",
    "bowling_rating",
    "form_rating"
]]

out.to_csv("player_features.csv", index=False)
print("✅ player_features.csv created")
```
